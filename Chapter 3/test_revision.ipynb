```python
def distance(x1, y1, x2, y2):
    return ((x2 -x1) **2 + (y2 - y1) ** 2) ** 0.5

x1, x2 = input("Enter the coordinates of the first point (x1 x2): ").split()
y1, y2 = input("Enter the coordinates of the second point (y1 y2): ").split()
x1, y1, x2, y2 = float(x1), float(y1), float(x2), float(y2)
print("The distance between the two points is:", distance(x1, y1, x2, y2))
```

```python
def is_palindrome(str1):
    if len(str1) < 2:
        print("The string is a palindrome")
        return
    if str1[0] != str1[-1]:
        print("The string is not a palindrome")
        return
    return is_palindrome(str1[1:-1])

str1 = input("Enter a string: ")
is_palindrome(str1)
```

```python
n_list = [10, 20, 30]

second_list = list(map(lambda x: x * 2, n_list))
third_list = list(map(lambda x: x * 3, n_list))

print("Twice the values:", second_list)
print("Thrice the values:", third_list)
```

```python
def cal_digit(n):
    def final(digit):
        return digit ** n
    return final

num_list = []
for num in range(1, 6):
    num_list.append(cal_digit(num))
    print(num_list [num - 1](num))
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __mul__(self, other):
        return self.x * other.x, self.y * other.y
    
    def __truediv__(self, other):
        return self.x / other.x, self.y / other.y
    
    def __str__(self):
        return "({:.2f}, {:.2f})".format(self.x, self.y)
    
v1 = Vector(30, 40)
v2 = Vector(10, 20)
v3 = v1 * v2
v4 = v1 / v2
print("Multiplication of vectors:", v3)
print("Division of vectors:", v4)
    
```
