```python
scores = [100, 90, 95, 90, 80, 70, 0, 80, 90, 90, 0, 90, 100, 75, 20, 30, 50, 90]

sub_groups = [scores[i: i + 3] for i in range(0, len(scores), 3)]

valid_scores = [sub_group for sub_group in sub_groups if 0 not in sub_group]

number_students = len(scores) // 3
number_students_valid = len(valid_scores)
print('The number of total students is {}.'.format(number_students))
print('The number of students with valid scores is {}.'.format(number_students_valid))
print(valid_scores)
```

```python
def func1(a):
    def func2():
        result1 = []
        for i in a:
            if i % 5 == 0:
                result1.append(i)
        return result1

    def func3():
        result2 = []
        for i in a:
            if i % 7 == 0:
                result2.append(i)
        return result2

    
    combined = func2() + func3()
    final_result = sorted(combined)

    print("result = {}".format(final_result))


lst = list(range(1, 101))
print("lst = {}".format(list(lst)))
func1(lst)
```
