## 1. Swap Operation

#### 1.1. Swap: exchange the value of two elements

```python
def swap1(S, x, y):
    t = S[x]
    S[x] = S[y]
    S[y] = t
S = [10, 20]
print(S)
swap1(S, 0, 1)
print(S)
```

#### 1.2. Swap in Python

```python
def swap1(S, x, y):
    S[x], S[y] = S[y], S[x]
S = [10, 20]
print(S)
swap1(S, 0, 1)
print(S)
```

## 2. Bubble Sort

#### 2.1. Implementing bubble Sort

```python
def bubblesort(S):
    n = len(S)
    for i in range(n):
        print(i)
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]
```

## 3. Selection Sort

#### 3.1. Implementing selection sort without swap

```python
def selectinsort1(S):
    R = []
    while len(S) > 0:
        print(R, S)
        smallest = S.index(min(S))
        R.append(S[smallest])
        S.pop(smallest)
    return R
```

#### 3.3. Implementing In-Place selection sort

```python
def selectinsort2(S):
    n = len(S)
    for i in range(n - 1):
        print(S)
        smallest = i
        for j in range(i + 1, n):
            if S[j] < S[smallest]:
                smallest = j
        S[i], S[smallest] = S[smallest], S[i]
```

## 4. Insertion Sort

#### 4.1. Implementing insertion sort with additional memory

```python
def insertionsort1(S):
    n = len(S)
    R = []
    while len(S) > 0:
        print(R, S)
        x = S.pop(0)
        j = len(R) - 1
        while j >= 0 and R[j] > x:
            j -= 1
        R.insert(j + 1, x)
    return R
```

```python
class HashTable:

    def __init__(self, size):
        self.size = size
        self.table = {}
        for i in range(size):
            self.table[i] = []
    
    def hash(self, key):
        return key % self.size
    
    def get(self, key):
        return self.table[self.hash(key)]
    
    def put(self, key, value):
        bucket = self.table[self.hash(key)]
        if value not in bucket:
            bucket.append(value)

table = HashTable(8)
book = "Alice in Wonderland"
key = sum(map(ord, book))
print(key, table.hash(key))


     
```

```python
merge_count = 0 

def merge(left, right):
    global merge_count
    merge_count += 1  
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(S):
    if len(S) <= 1:
        return S
    mid = len(S) // 2
    left = merge_sort(S[:mid])
    right = merge_sort(S[mid:])
    return merge(left, right)

S = [6, 2, 11, 7, 5, 4, 8, 16, 10, 3]
sorted_S = merge_sort(S)
print(sorted_S)
print("Merge function was called", merge_count, "times")
```
